```python
# Calculate Mean

mean = torch.zeros(3)
std = torch.zeros(3)
total_images = 0


for images , labels in train_loader:

    #reshape
    images = images.view(images.size(0),images.size(1) , -1)

    #compute mean 
    batch_mean = images.mean(2)
    mean += batch_mean.sum(0)

    #compute std
    batch_std = images.std(2)
    std += batch_std.sum(0)

    total_images += images.size(0)
    

mean = mean/total_images
std = std/total_images

print("Mean: ",mean)
print("Std :", std)
```

```python
#Normalization


transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean,std)
    
])

train_dataset = datasets.ImageFolder(train_dir,transform = transform)

train_loader = DataLoader(train_dataset , batch_size=32 , shuffle=True)


images, _ = next(iter(train_loader))

print("Mean after normalize:", images.mean(dim=[0,2,3]))
print("Std after normalize:", images.std(dim=[0,2,3]))
```
